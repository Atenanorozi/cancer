### سوال 12

نمودار تعداد کلاس‌های متغیر هدف را رسم کنید.

```python
plt.figure(figsize=(6,4))

sns.countplot(data=df,x="Level")

plt.title("Distribution of Lung Cancer Level")
plt.show()
```

بیشترین فراوانی را دارد و توزیع کلاس‌ها تقریباً متعادل است. high نمودار نشان می‌دهد کلاس 

# بخش سوم : تحلیل ویژگی ها

### سوال 13

توزیع سنی بیماران را بررسی کنید

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=15, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

این نمودار نشان می‌دهد سن بیماران بیشتر در بازه 25 تا 50 سال قرار دارد و توزیع سن تقریباً نرمال است.
