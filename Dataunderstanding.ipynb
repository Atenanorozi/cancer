# بخش سوم : تحلیل ویژگی ها

### سوال 13

توزیع سنی بیماران را بررسی کنید

```python
plt.figure(figsize=(8,5))

sns.histplot(df["Age"], bins=15, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

این نمودار نشان می‌دهد سن بیماران بیشتر در بازه 25 تا 50 سال قرار دارد و توزیع سن تقریباً نرمال است.

### سوال 14

میانگین، میانه، کمترین و بیشترین سن را محاسبه کنید

```python
mean_age = df["Age"].mean()
median_age = df["Age"].median()
min_age = df["Age"].min()
max_age = df["Age"].max()


print("mean age :", mean_age)
print("median age :", median_age)
print("minimum age :", min_age)
print("maximum age :", max_age)
```

میانگین سن بیماران حدود 37 سال است و سن افراد بین 14 تا 73 سال قرار دارد.

### سوال 15

```python
df["Gender"].value_counts()
```

```python
plt.figure(figsize=(5,4))

sns.countplot(data=df, x="Gender")

plt.title("Gender Distribution")
plt.show()
```

این نمودار تعداد افراد هر جنسیت را نمایش می‌دهد.

### سوال 16

#  بخش ششم : همبستگی

### سوال 28

ماتریس همبستگی بین ویژگی‌های عددی را محاسبه کنید.

```python
corr_matrix = df.corr(numeric_only=True)
corr_matrix
```

این ماتریس میزان همبستگی بین تمام ویژگی‌های عددی را نمایش می‌دهد. ضرایب نزدیک به 1 نشان‌دهنده همبستگی مثبت قوی و ضرایب نزدیک به -1 نشان‌دهنده همبستگی منفی قوی هستند.


با بررسی ماتریس همبستگی مشاهده می‌شود که برخی ویژگی‌ها مانند Smoking، Passive Smoker، Air Pollution و Genetic Risk همبستگی بیشتری با سایر ویژگی‌ها دارند.

### سوال 29  

Heatmap ماتریس همبستگی را رسم کنید.

```python
plt.figure(figsize=(18,12))

sns.heatmap(corr_matrix,
            annot=True,
            cmap="coolwarm",
            fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

با استفاده از Heatmap می‌توان ویژگی‌هایی را که بیشترین یا کمترین همبستگی را با یکدیگر دارند، به‌راحتی شناسایی کرد.

### سوال 30

نمودار Scatter بین سن و میزان سیگار کشیدن را رسم کنید

```python
plt.figure(figsize=(7,5))

sns.scatterplot(data=df,
                x="Age",
                y="Smoking",
                hue="Level")

plt.title("Age vs Smoking")
plt.show()
```

این نمودار رابطه بین سن و میزان سیگار کشیدن را نشان می‌دهد.

### سوال 31

Pairplot برای چند ویژگی مهم رسم کنید.

```python
important_feature = ["Age", "Smoking", "Air Pollution", "Genetic Risk", "Level"]
sns.pairplot(
    df[important_feature],
    hue="Level"
)

plt.show()
```

Pairplot ارتباط بین چند ویژگی مهم را به‌صورت هم‌زمان نمایش می‌دهد و به تشخیص الگوهای داده کمک می‌کند

### سوال 32

میانگین ویژگی‌های مهم را بر اساس Level محاسبه کنید.

```python
df.groupby("Level")[["Age",
                     "Smoking",
                     "Air Pollution",
                     "Genetic Risk"]].mean()
```

این جدول میانگین ویژگی‌های مهم را برای هر سطح خطر (Low، Medium و High) نمایش می‌دهد و امکان مقایسه بین گروه‌ها را فراهم می‌کند.

### سوال 33

کدام ویژگی‌ها بیشترین تأثیر را بر سطح سرطان ریه دارند؟

```python
df.groupby("Level").mean(numeric_only=True).T
```

با مقایسه میانگین ویژگی‌ها در سه کلاس، می‌توان مشاهده کرد که ویژگی‌هایی مانند Smoking، Air Pollution، Genetic Risk و Passive Smoker معمولاً در کلاس High مقادیر بیشتری دارند.
